```python
import pandas as pd
```

```python
SOURCE = "Pedestrian_Counting_System_Monthly_counts_per_hour_may_2009_to_14_dec_2022.csv"
YEAR = 2018
MONTH = "March"
```

```python
cols = ["Year", "Month", "Mdate", "Time", "Sensor_ID", "Sensor_Name", "Hourly_Counts"]
```

```python
parts = []
for chunk in pd.read_csv(SOURCE, usecols=cols, chunksize=200_000, thousands=","):
    parts.append(chunk[(chunk["Year"] == YEAR) & (chunk["Month"] == MONTH)])
 
month = pd.concat(parts)
```

```python
month["Sensor_Name"] = month["Sensor_Name"].str.strip()
 
out = f"melbourne_{YEAR}_{pd.to_datetime(MONTH, format='%B').month:02d}.csv"
month[["Sensor_ID", "Sensor_Name", "Mdate", "Time", "Hourly_Counts"]].to_csv(out, index=False)
 
print(f"{out}: {len(month):,} rows, {month['Sensor_ID'].nunique()} sensors")
```
